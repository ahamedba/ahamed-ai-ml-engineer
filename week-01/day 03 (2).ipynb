Step 1: Understand the Business Problem

A small retail business stores its daily sales transactions in a CSV file.

The store manager wants a weekly sales report that shows total revenue, total units sold, average transaction value, and revenue generated by each product.

The application should also check whether the sales data is correct and complete.

Based on the calculated revenue, the application should determine whether the business achieved its weekly revenue target.

Finally, the application should generate reports so the store manager can understand the business performance and make better decisions.

```python
def load_csv(file_path):
    rows = []

    with open(file_path, mode="r", newline="", encoding="utf-8") as file:
        reader = csv.DictReader(file)

        for row in reader:
            rows.append(row)

    return rows
```
