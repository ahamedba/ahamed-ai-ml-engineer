```python
from pathlib import Path

file_path = Path("week-01/day-02/exercises/loops_demo.py")

code = '''
daily_sales = [1200, 1500, 900, 1800]

print("FOR LOOP")

for sale in daily_sales:
    print(sale)


print("\\nWHILE LOOP")

index = 0

while index < len(daily_sales):
    print(daily_sales[index])
    index += 1


print("\\nRANGE EXAMPLE")

for number in range(1, 6):
    print(number)


print("\\nBREAK EXAMPLE")

for sale in daily_sales:
    if sale == 900:
        break
    print(sale)


print("\\nCONTINUE EXAMPLE")

for sale in daily_sales:
    if sale == 900:
        continue
    print(sale)
'''

file_path.write_text(code.strip() + "\n", encoding="utf-8")

print("Created:", file_path)
```

```python
from pathlib import Path

file_path = Path("week-01/day-02/exercises/functions_demo.py")

code = '''
def calculate_total(quantity, unit_price):
    return quantity * unit_price


order_total = calculate_total(4, 125.50)
print(f"Order total: {order_total:.2f}")


zero_total = calculate_total(0, 125.50)
print(f"Zero quantity total: {zero_total:.2f}")
'''

file_path.write_text(code.strip() + "\n", encoding="utf-8")

print("Created:", file_path)
```
