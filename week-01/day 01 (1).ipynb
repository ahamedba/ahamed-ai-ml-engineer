Definitions in my own words
1. Artificial Intelligence (AI): AI is a technology that enables computers or machines to perform tasks that normally require human intelligence, such as decision-making, problem-solving, understanding language and recognizing patterns.

2. Machine Learning (ML): ML is a subset of AI where computers learn from historical data, identify patterns and make predictions or decisions without being explicitly programmed for every task.

3. Deep Learning: Deep Learning is a subset of Machine Learning that uses multi-layered neural networks to learn complex patterns from large amounts of data.

4. Generative AI: Generative AI is a type of AI that can create new content, such as text, images, videos, audio and computer code, based on what it has learned from data.

4. Checkpoint questions
Q1. Which one is the widest concept?

Answer: Artificial Intelligence (AI) is the widest concept because it includes Machine Learning, Deep Learning and other approaches such as rule-based systems.

Q2. What is the main difference between AI and ML?

Answer: AI is the broader field of making machines intelligent, whereas ML is an approach that allows machines to learn patterns from data.

Q3. Can AI exist without ML?

Answer: Yes. Rule-based systems can perform intelligent tasks using predefined instructions without learning from data.

Rule-Based Programming vs Machine Learning
1. Retail example suited for Rule-Based Programming
Example: Discount eligibility

A retail company offers a 10% discount to customers whose purchase value is above ₹5,000.

Rule:
order_value = 6000if order_value > 5000:    print("Customer is eligible for 10% discount")else:    print("Customer is not eligible for discount")



Why Rule-Based Programming?
- The condition is already known.
- The rule is simple and consistent.
- We don't need historical data to make the decision.
- The same rule can be applied to every customer.
2. Retail example suited for Machine Learning
Example: Customer Purchase Prediction

A retail company wants to predict which customers are likely to purchase a product in the future.

Input data (features):
- Customer age
- Previous purchase history
- Total spending
- Number of website visits
- Product preferences
Target: Whether the customer makes a purchase (Yes/No).

Why Machine Learning?
- Customer purchasing behaviour is complex.
- There may not be a fixed rule to identify every potential buyer.
- ML can learn patterns from historical customer data.
- It can predict which customers are more likely to purchase.

4. Checkpoint – My Answer
Q: Give one retail example that suits rules and one that suits ML. Say why.

Answer:
1. Rule-Based: Applying a discount when a customer's purchase value exceeds ₹5,000. This is suitable because the condition is clear, fixed and easy to implement using an IF-ELSE statement.
2. Machine Learning: Predicting whether a customer will purchase a product based on previous shopping behaviour. This is suitable because customer behaviour involves multiple factors, and ML can identify patterns from historical data to make predictions.

Machine Learning Project Lifecycle – 8 Stages
Why it matters: A Machine Learning project is not just about training a model. It involves understanding the business problem, preparing data, building a model, evaluating results and monitoring its performance.
1. Eight stages of the ML lifecycle
1. Define the Business Problem

Understand the business requirement and identify what problem needs to be solved.

Example: A retail store wants to predict next week's product sales to avoid overstocking and stock shortages.

2. Collect and Understand Data

Gather relevant historical data and understand its structure, quality and patterns.

Example: Collect sales records, product details, prices, discounts and seasonal information.

3. Prepare Data and Features

Clean missing or incorrect values, remove duplicates and select useful features for the model.

Example: Prepare product prices, previous sales quantities, discounts and holiday information.

4. Set a Baseline

Create a simple starting prediction to compare against the ML model.

Example: Predict next week's sales using the average sales of the previous four weeks.

5. Train the Model

Feed prepared training data into an ML algorithm so it can learn patterns.

Example: Train a Linear Regression model using historical retail sales data.

6. Evaluate the Model

Test the model using unseen data and measure its performance.

Example: Compare predicted sales with actual sales using MAE or RMSE.

7. Deploy the Model

Make the trained model available for real-world business use.

Example: Integrate the sales prediction model into a retail dashboard.

8. Monitor the Model

Continuously track model performance and update it when business conditions or data patterns change.

Example: Monitor prediction accuracy when customer demand changes during festival seasons.

3. Checkpoint Questions
Q1. Why does the first stage come before any data work?

Answer:

The first stage is defining the business problem because we must understand what the business wants to achieve before collecting data.

It helps us identify:
- What problem needs to be solved.
- What data is required.
- What features and target variables are needed.
- Whether Machine Learning is the right solution.
Without understanding the business problem, we may collect irrelevant data and build a model that does not solve the actual business requirement.

Q2. Can every business problem be solved with ML? Explain with an example.

Answer:

No, every business problem cannot be solved using Machine Learning.

Some problems have fixed rules and simple conditions, where traditional programming is more suitable.

Example: Retail discount calculation

A store provides a 10% discount when a customer's purchase value exceeds ₹5,000.

This can be solved using a simple IF-ELSE condition instead of Machine Learning because the business rule is already defined.

However, predicting which customers are likely to purchase products is better suited to Machine Learning because customer behaviour involves complex patterns.
